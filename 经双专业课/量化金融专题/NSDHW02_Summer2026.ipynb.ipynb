## Ito isometry

### 1. (20pts)
Find the variance of the following random variables by applying the Ito isometry 

$$
\Eof{\left(\int_a^b \varphi_s dB_s\right)^2} = \int_a^b \Eof{\varphi_s^2}ds
$$

- (a) $X = \dint_0^t B_s^2 dB_s$. In this case, $\varphi_s = B_s^2$.
- (b) $X = \int_a^b \sqrt t B_t dB_t$. In this case, $\varphi_t = \sqrt t B_t$.

### <font color=blue> Solution 1(a). </font>
由于被积函数 $\varphi_s = B_s^2$ 是适应的（仅依赖于当前时刻的$B_s$），且满足平方可积条件：
$$
\int_0^t \E[\varphi_s^2] ds = \int_0^t \E[B_s^4] ds = \int_0^t 3s^2 ds < \infty
$$
根据伊藤积分的鞅性质，故
$$
\Eof{X}=\Eof{\int_0^tB_s^2dB_s}=0
$$
根据Ito isometry
$$
\Eof{X^2}=\Eof{\left(\int_0^tB_s^2dB_s \right) ^2}=\int_0^t\Eof{\left( B_s^2\right) ^2}ds=\int_0^t\Eof{B_s^4}ds
$$
由于$B_t$是标准布朗运动，故
$$
B_s \sim \mathcal{N}(0,s)
$$
其四阶矩
$$
\Eof{B_s^4}=3\left(\sqrt{s}\right)^4=3s^2
$$
故
$$
\Eof{X^2}=\int_0^t\Eof{B_s^4}ds=\int_0^t3s^2ds=\left[s^3\right]\big|_{s=0}^t=t^3
$$
因而X的方差
$$
\operatorname{Var}[X]=\Eof{X^2}-\left(\Eof{X}\right)^2=t^3-0=t^3
$$

### <font color=blue> Solution 1(b). </font>
由于被积函数 $\varphi_t = \sqrt{t}B_t$ 是适应的（仅依赖于当前时刻的$B_t$和确定性函数$\sqrt{t}$），且满足平方可积条件：
$$
\int_a^b \E[\varphi_t^2] dt = \int_a^b \E[(\sqrt{t} B_t)^2] dt = \int_a^b t \cdot \E[B_t^2] dt = \int_a^b t^2 dt < \infty
$$
根据伊藤积分的鞅性质，故
$$
\Eof{X} = \Eof{\int_a^b \sqrt{t} B_t \, dB_t} = 0
$$
根据Ito isometry
$$
\Eof{X^2} = \Eof{\left(\int_a^b \sqrt{t} B_t \, dB_t\right)^2} = \int_a^b \Eof{(\sqrt{t} B_t)^2} dt = \int_a^b t \cdot \Eof{B_t^2} dt
$$
由于$B_t$是标准布朗运动，故
$$
B_t \sim \mathcal{N}(0,t)
$$
其二阶矩
$$
\Eof{B_t^2} = \left(\sqrt{t}\right)^2=t
$$
故
$$
\Eof{X^2} = \int_a^b t \cdot \Eof{B_t^2} dt= \int_a^b t \cdot t \, dt = \int_a^b t^2 dt = \left[ \frac{t^3}{3} \right]\Big|_{t=a}^b = \frac{b^3 - a^3}{3}
$$
因而X的方差
$$
\operatorname{Var}[X]=\Eof{X^2}-\left(\Eof{X}\right)^2=\frac{b^3 - a^3}{3}-0=\frac{b^3 - a^3}{3}
$$

## Apply Ito's formula to evaluate stochastic integral

### 2. (20pts)
Apply Ito's formula to evaluate the following stochastic integrals.

- (a) $\dint_0^t B_s e^{B_s} dB_s$ 
- (b) $\dint_0^t \frac 1{1 + B_s^2} dB_s$ 

### <font color=blue> Solution 2(a). </font>
考虑函数
$$
F(x)=(x-1)e^x
$$
其前两阶导数
$$
F'(x)=xe^x
$$
$$
F''(x)=(x+1)e^x
$$
因而，对$F(B_s)=(B_s-1)e^{B_s}$使用伊藤公式得
$$
d\left(F(B_s)\right)=F'(B_s)dB_s+\frac{1}{2}F''(B_s)ds
$$
即
$$
d\left[(B_s-1)e^{B_s}\right]=B_se^{B_s}dB_s+\frac{1}{2}(B_s+1)e^{B_s}ds
$$
带入题中积分得
$$
\int_0^tB_se^{B_s}dB_s
=\int_0^td\left[(B_s-1)e^{B_s}\right]-\int_0^t\frac{1}{2}(B_s+1)e^{B_s}ds \\
=\left[(B_s-1)e^{B_s}\right]\Big|_{s=0}^t-\frac{1}{2}\int_0^t(B_s+1)e^{B_s}ds \\
=(B_t-1)e^{B_t}-(B_0-1)e^{B_0}-\frac{1}{2}\int_0^t(B_s+1)e^{B_s}ds \\
\xlongequal{标准布朗运动B_0=0}(B_t-1)e^{B_t}-(0-1)e^{0}-\frac{1}{2}\int_0^t(B_s+1)e^{B_s}ds \\
=(B_t-1)e^{B_t}+1-\frac{1}{2}\int_0^t(B_s+1)e^{B_s}ds
$$
因而结果为：
$$
\int_0^t B_s e^{B_s} dB_s = (B_t - 1)e^{B_t} + 1 - \frac{1}{2} \int_0^t (B_s + 1)e^{B_s} ds
$$

### <font color=blue> Solution 2(b). </font>
考虑函数
$$
F(x)=\arctan{x}
$$
其前两阶导数
$$
F'(x)=\frac{1}{1+x^2}
$$
$$
F''(x)=-\frac{2x}{(1+x^2)^2}
$$
因而，对$F(B_s)=\arctan{B_s}$使用伊藤公式得
$$
d\left(F(B_s)\right)=F'(B_s)dB_s+\frac{1}{2}F''(B_s)ds
$$
即
$$
d\left[\arctan{B_s}\right]=\frac{1}{1+{B_s}^2}dB_s-\frac{1}{2}\cdot\frac{2{B_s}}{(1+{B_s}^2)^2}ds=\frac{1}{1+{B_s}^2}dB_s-\frac{B_s}{(1+{B_s}^2)^2}ds
$$
带入题中积分得
$$
\int_0^t\frac{1}{1+{B_s}^2}dB_s
=\int_0^td\left[\arctan{B_s}\right]+\int_0^t\frac{{B_s}}{(1+{B_s}^2)^2}ds \\
=\left[\arctan{B_s}\right]\Big|_{s=0}^t+\int_0^t\frac{{B_s}}{(1+{B_s}^2)^2}ds \\
=\arctan{B_t}-\arctan{B_0}+\int_0^t\frac{{B_s}}{(1+{B_s}^2)^2}ds \\
\xlongequal{标准布朗运动B_0=0}\arctan{B_t}-\arctan{0}+\int_0^t\frac{{B_s}}{(1+{B_s}^2)^2}ds \\
=\arctan{B_t}+\int_0^t\frac{{B_s}}{(1+{B_s}^2)^2}ds
$$
因而结果为：
$$
\int_0^t\frac{1}{1+{B_s}^2}dB_s = \arctan{B_t}+\int_0^t\frac{{B_s}}{(1+{B_s}^2)^2}ds
$$

## Apply Ito's formula to solve SDE

### 3. (10pts)

Let $X_t$ satisfy the SDE for Ornstein-Uhlenbeck process

$$
dX_t = \lambda(m - X_t) dt + \sigma dB_t, \quad X_0 = x > 0,
$$

where $\lambda > 0$, $m$ and $\sigma$ are constants. 

Apply Ito's formula to $e^{\lambda t}X_t$ and show that the solution is given by

$$
X_t = e^{-\lambda t} x + \left(1 - e^{-\lambda t} \right) m + \sigma \int_0^t e^{-\lambda(t - s)} dW_s.
$$

### <font color=blue> Solution 3. </font>
设$f(t,x)=e^{\lambda t}x$

则前两阶偏导数
$$
f_t(t,x)=\frac{\partial}{\partial t}\left(e^{\lambda t}x\right)=\lambda x e^{\lambda t}
$$
$$
f_x(t,x)=\frac{\partial}{\partial x}\left(e^{\lambda t}x\right)=e^{\lambda t}
$$
$$
f_{xx}(t,x)=\frac{\partial^2}{\partial x^2}\left(e^{\lambda t}x\right)=0
$$
对$f(t,X_t)=e^{\lambda t}X_t$使用伊藤公式得
$$
d\left[f(t,X_t)\right]=f_x(t,X_t)dX_t+\left[f_t(t,X_t)+\frac{1}{2}f_{xx}(t,X_t)\right]dt
$$
即
$$
d\left[e^{\lambda t}X_t\right]=e^{\lambda t}dX_t+ \lambda X_t e^{\lambda t}dt
$$
带入$dX_t = \lambda(m - X_t) dt + \sigma dW_t$得
$$
d\left[e^{\lambda t}X_t\right]
=\lambda(m - X_t)e^{\lambda t}dt+\sigma e^{\lambda t}dW_t +\lambda X_t e^{\lambda t}dt 
=\lambda m e^{\lambda t}dt+\sigma e^{\lambda t}dW_t
$$
等式两侧从0到t积分得
$$
\int_0^td\left[e^{\lambda s}X_s\right]=\int_0^t\lambda m e^{\lambda s}ds+\int_0^t\sigma e^{\lambda s}dW_s
$$
$$
\left[e^{\lambda s}X_s\right]\Big|_{s=0}^t
=\left[me^{\lambda s}\right]\Big|_{s=0}^t+\int_0^t\sigma e^{\lambda s}dW_s
$$
$$
e^{\lambda t}X_t-X_0=me^{\lambda t}-m+\int_0^t\sigma e^{\lambda s}dW_s
$$
带入$X_0=x$得
$$
e^{\lambda t}X_t-x=me^{\lambda t}-m+\int_0^t\sigma e^{\lambda s}dW_s
$$
$$
e^{\lambda t}X_t=x + (e^{\lambda t}-1)m+\sigma\int_0^t e^{\lambda s}dW_s
$$
即
$$
X_t = e^{-\lambda t} x + \left(1 - e^{-\lambda t} \right) m + \sigma \int_0^t e^{-\lambda(t - s)} dW_s
$$

## Simulating the Black-Scholes model

### 4. (20pts)

With 100 steps within the time interval $[0,1]$, simulate 20,000 sample paths for the Black-Scholes model 

$$
dX_t = r X_t dt + \sigma X_t dB_t
$$

by the Euler-Maruyama scheme with $X_0 = 100$. The parameters are given by $r = 2\%$ and $\sigma = 25\%$.

- (a) Plot the histograms of $X_t$ for $t = 0.4, 1$.
- (b) Evaluate the expectation $ e^{-rT} \Eof{g(X_T)} $ by taking sample mean from using the simulated samples in (a), where $T = 1$ and $g$ is given by 
$$
g(x) = \left\{\begin{array}{ll}
5 & \text{ if } x > 101, \\
-4 & \text{ if } x \leq 101.
\end{array}\right.
$$
Plot the histogram of $g(X_1)$. 

Note: In `python`, the function $g$ can be coded as `5*(x>101) - 4*(x<=101)`. 

```python
# as always, import required modules and functions
import numpy as np
from numpy import sqrt, log, exp
import matplotlib.pyplot as plt
import scipy.stats as ss
from scipy.stats import norm
import seaborn as sns
```

```python
#(4a)
# set random seed
np.random.seed(1609)

# 模拟 Black-Scholes 模型：dX_t = r X_t dt + sigma X_t dB_t

# 参数设置
r, sigma = 0.02, 0.25
X0, T = 100, 1

# number of paths and number of time steps
n_sim, n_steps = 20_000, 100

# initialize X
X = np.ones([n_sim, n_steps+1])*X0

# Euler-Maruyama scheme
dt = T/n_steps
for i in range(n_steps):
    db = norm.rvs(size=n_sim)
    db = db - db.mean()
    db = db/db.std()
    X[:, i+1] = X[:, i] + r * X[:, i] * dt + sigma * X[:, i] * np.sqrt(dt) * db

# plot
plt.figure(figsize=(12, 8))
t = np.arange(0, T+dt, dt)

# histograms at different times
time = 0.4 # 0.5, 0.8
sns.histplot(X[:,t==time], bins=30, stat='density', label=f'$t={time}$', element='step')
sns.histplot(X[:,-1], bins=30, stat='density', label=f'$t={T}$', color='lightgreen', element='step')
plt.legend()
plt.title('Histograms of $X_t$ at t=0.4 and t=1');
```

```python
#(4b)
# Define payoff function g
g = 5 * (X[:, -1] > 101) - 4 * (X[:, -1] <= 101)

# Compute discounted expectation
discounted_mean = np.exp(-r * T) * np.mean(g)
print(f"e^(-rT) * E[g(X_T)] = {discounted_mean:.4f}")
```

```python
# histogram of g(X1)
plt.figure(figsize=(8, 5))
sns.histplot(g, bins=2, stat='probability', discrete=True)
plt.xlabel('$g(X_1)$')
plt.xticks([-4, 5])
plt.ylabel('Frequency')
plt.title('Histogram of $g(X_1)$')

counts = np.bincount((g == 5).astype(int))
probs = counts / len(g)
for i, v in enumerate(probs):
    x_pos = -4 if i == 0 else 5
    plt.text(x_pos, v
    , f'{v:.3f}', ha='center', va='bottom')
plt.show()
```
