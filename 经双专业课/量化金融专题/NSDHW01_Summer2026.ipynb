### 1. (5 points)
Let $X_t = k B_{t/k^2}$, where $k \neq 0$ is constant. Show by definition that $X_t$ is also a standard Brownian motion. 

### <font color=blue> Solution 1. </font>
设$B_t$是定义在滤子概率空间$(\Omega,\mathcal{F}_t,\mathbb{P})$上的标准布朗运动 \
若令$\mathcal{G}_t=\mathcal{F}_{t/{k^2}}$ \
则$X_t=kB_{t/k^2}$是定义在滤子概率空间$(\Omega,\mathcal{G}_{t},\mathbb{P})$上的随机过程 \
下面验证$X_t$满足标准布朗运动的四条定义 \
**①初值条件：**
$$
\mathbb{P}[\omega: X_0(\omega)=0]=\mathbb{P}[\omega:kB_{0/k^2}(\omega)=0]=\mathbb{P}[\omega:B_0(\omega)=0]=1
$$
**②正态增量：**
对任意$0 \leq s < t$，随机变量$X_t-X_s$对于任意的 $a < b $满足 \
若$k>0$，则
$$
\mathbb{P}[a \leq X_t - X_s \leq b]=\mathbb{P}[a \leq kB_{t/k^2} - kB_{s/k^2} \leq b]\xlongequal{k>0}
\mathbb{P}[\frac{a}{k} \leq B_{t/k^2} - B_{s/k^2} \leq \frac{b}{k}] \\
\xlongequal{B_t标准布朗运动}
\frac{1}{\sqrt{2\pi(t/k^2-s/k^2)}}\int_{a/k}^{b/k}e^{-\frac{x^2}{2(t/k^2-s/k^2)}}dx \\
\xlongequal{k > 0}\frac{k}{\sqrt{2\pi(t-s)}}\int_{a/k}^{b/k}e^{-\frac{x^2k^2}{2(t-s)}}dx \\
\xlongequal{令y=kx}\frac{1}{\sqrt{2\pi(t-s)}}\int_{a}^{b}e^{-\frac{y^2}{2(t-s)}}dy
$$
若$k<0$，则
$$
\mathbb{P}[a \leq X_t - X_s \leq b]=\mathbb{P}[a \leq kB_{t/k^2} - kB_{s/k^2} \leq b]\xlongequal{k<0}
\mathbb{P}[\frac{b}{k} \leq B_{t/k^2} - B_{s/k^2} \leq \frac{a}{k}] \\
\xlongequal{B_t标准布朗运动}
\frac{1}{\sqrt{2\pi(t/k^2-s/k^2)}}\int_{b/k}^{a/k}e^{-\frac{x^2}{2(t/k^2-s/k^2)}}dx \\
\xlongequal{k < 0}\frac{-k}{\sqrt{2\pi(t-s)}}\int_{b/k}^{a/k}e^{-\frac{x^2k^2}{2(t-s)}}dx \\
\xlongequal{令y=kx}\frac{1}{\sqrt{2\pi(t-s)}}\int_{a}^{b}e^{-\frac{y^2}{2(t-s)}}dy
$$
综上，
$$
\mathbb{P}[a \leq X_t - X_s \leq b]=\frac{1}{\sqrt{2\pi(t-s)}}\int_{a}^{b}e^{-\frac{y^2}{2(t-s)}}dy
$$
即，
$$
X_t - X_s \sim \mathcal{N}(0, t-s)
$$
**③独立增量**
$$
对于任意 0 \leq t_1 < t_2 < \cdots < t_n  \\
\because k \neq 0，k^2 > 0 \\
\therefore  0 \leq \frac{t_1}{k^2} < \frac{t_2}{k^2} < \cdots < \frac{t_n}{k^2} \\
又 \because B_t 是标准布朗运动  \\
\therefore B_{t_1/k^2},B_{t_2/k^2}-B_{t_1/k^2},\cdots,B_{t_n/k^2}-B_{t_{n-1}/k^2}相互独立 \\
又 \because k \neq 0 且为常数 \\
\therefore kB_{t_1/k^2},kB_{t_2/k^2}-kB_{t_1/k^2},\cdots,kB_{t_n/k^2}-kB_{t_{n-1}/k^2}相互独立 \\
即X_{t_1},X_{t_2}-X_{t_1},\cdots,X_{t_n}-X_{t_{n-1}}相互独立
$$
**④路径连续性**
$$
\because B_t 是标准布朗运动 \\
\therefore \mathbb{P}[\omega: B_t(\omega)是连续函数]=1 \\
对于任意固定样本点\omega， \\
若B_t(\omega)作为t的函数是连续的 \\
\because 映射 t \mapsto t/k^2 是连续的 \\
\therefore 连续函数的复合 t \mapsto B_{t/k^2}(\omega) 是连续的 \\
\therefore t \mapsto kB_{t/k^2}(\omega) 是连续的 \\
即X_t(\omega)作为t的函数时连续的 \\
\therefore 对于几乎所有的\omega，映射 t \mapsto X_t(\omega)都是连续函数 \\
即 \mathbb{P}[\omega: X_t(\omega)是连续函数]=1
$$
由**①②③④**，\
$X_t$是定义在滤子概率空间$(\Omega,\mathcal{G}_t,\mathbb{P})$上的标准布朗运动

## Donsker's invariance principle

Recall the Donsker's invariance principle: 


Suppose $\{X_i\}_{i=1}^\infty$ is an iid sequence of random variables with mean 0 and and variance 1. Let $S_n = \sum_{i=1}^n X_i$. Define the function $\mathfrak{S}_n$ of $t$ by 

$$
  \mathfrak{S}_n(t) = \frac1{\sqrt n}\sum_{i=1}^n\left[S_{i-1} + n \left( t-\frac{i-1}n \right) X_i \right]\,\mathbf{1}_{\left(\frac{i-1}n,\frac in\right]}(t) .
$$

In fact, $\mathfrak{S}_n$ is simply the linear interpolation of the scaled random walk $\left\{\frac{S_1}{\sqrt n}, \frac{S_2}{\sqrt n}, \cdots, \frac{S_n}{\sqrt n} \right\}$.

Then, $\mathfrak{S}_n\Longrightarrow W$ as $n\to\infty$, where $W$ denotes a standard Brownian motion. 

In other words, as $n\to\infty$, the linearly interpolated scaled random walk $\mathfrak{S}_n$ converges weakly to a standard Brownian motion. 

### 2. (10pts)
In this problem, we show, numerically by simulation, the Donsker's invariance principle with the iid sequence $X_i$ being 

- (a) $X_i \sim t_3$, i.e., the $t$ distribution with degrees of freedom 3;
- (b) $X_i \sim \exp(1)$, i.e., exponential distribution with mean 1.

Consider the Brownian motion in the time interval $[0,2]$, use $100$ time steps and generate $15,000$ sample paths to reproduce the plots for sample mean, sample standard deviation, and the histograms at $t=1$ and $t=2$. To begin with, please set the random seed by including ``numpy.random.seed(9527)`` on top of your code. 

```python
# as always, import required modules and functions
import numpy as np
from numpy import sqrt, log, exp
import matplotlib.pyplot as plt
import scipy.stats as ss
from scipy.stats import norm
import seaborn as sns
```

### <font color=blue> Solution 2(a). </font>

```python
# insert python code for (2a) in this cell
np.random.seed(9527)
# Simulate BM by using the Donsker's invariance principle
# number of samples, number of steps, terminal time
n_sim, n_steps, T = 15_000, 100, 2  
dt = T/n_steps

# initialize the Brownians
B = np.zeros([n_sim, n_steps+1])

# simulation step
for i in range(n_steps):
    dB = ss.t.rvs(df=3, size=n_sim) / np.sqrt(3)
    dB = np.sqrt(dt)*dB
    B[:,i+1] = B[:,i] + dB 


# plot
t = np.arange(0, T+dt, dt)
#t = np.linspace(0, T, n_steps+1)
plt.figure(figsize=(10, 8))

# sample path
path = np.random.choice(n_sim)
plt.subplot(2, 2, 1)
plt.plot(t, B[path, :], label=f'sample path #{path}')
plt.xlabel('$t$')
plt.ylabel('$B_t$ sample path')
plt.legend()

# histograms
plt.subplot(2, 2, 2)
time = 1
sns.histplot(B[:,t==time], bins=50, stat='density', label=f'$t={time}$')
sns.histplot(B[:,-1], bins=50, stat='density', color='orange', label=f'$t={T}$')
x = np.linspace(B[:,-1].min(), B[:,-1].max(), 100)
plt.plot(x, norm.pdf(x, scale=sqrt(time)), 'y--', lw=1, label=f'normal pdf  $t={time}$')
plt.plot(x, norm.pdf(x, scale=sqrt(T)), 'r--', lw=1, label=f'normal pdf  $t={T}$')
plt.xlabel('$B$')
plt.legend();

# sample mean
plt.subplot(2, 2, 3)
plt.plot(t, B.mean(axis=0), label='sample mean')
plt.hlines(y=0, xmin=0, xmax=T, color='red', ls='dotted', label='theoretical mean (0)')
plt.xlabel('$t$')
plt.legend()

# sampel standard deviation
plt.subplot(2, 2, 4)
plt.plot(t, B.std(axis=0), label='sample std')
plt.plot(t, np.sqrt(t), 'r--', label='sqrt t')
plt.xlabel('$t$')
plt.legend();
```

Alter the `python` code provided in the lecture to simulate the process $X_t$ given by

X_t = 1 + 3t(2-t) + B_t,

for $0 \leq t \leq 2$. Use $200$ time steps and generate $20,000$ sample paths, plot sample mean, sample standard deviation (as functions of $t$), and the histograms at $t = 1$ and $t=2$.  To begin with, set `numpy.random.seed(1732)`. 

### <font color=blue> Solution 3. </font>

```python
# insert python code for (3) in this cell
np.random.seed(1732)

# Simulate by using the Donsker's invariance principle
# number of samples, number of steps, terminal time
n_sim, n_steps, T = 20_000, 200, 2  
dt = T/n_steps

# initialize the Brownians
X = np.zeros([n_sim, n_steps+1])
X[:, 0] = 1.0 

# simulation step
for i in range(n_steps):
    dB = norm.rvs(size=n_sim)
    dB = np.sqrt(dt) * dB
    t_i = i * dt
    drift = 6 - 6 * t_i
    X[:, i + 1] = X[:, i] + drift * dt + dB


# plot
t = np.arange(0, T+dt, dt)
#t = np.linspace(0, T, n_steps+1)
plt.figure(figsize=(10, 8))

# sample path
path = np.random.choice(n_sim)
plt.subplot(2, 2, 1)
plt.plot(t, X[path, :], label=f'sample path #{path}')
plt.xlabel('$t$')
plt.ylabel('$X_t$ sample path')
plt.legend()

# histograms
plt.subplot(2, 2, 2)
time = 1
sns.histplot(X[:,t==time], bins=50, stat='density', label=f'$t={time}$')
sns.histplot(X[:,-1], bins=50, stat='density', color='orange', label=f'$t={T}$')
x1 = np.linspace(X[:, t == time].min(), X[:, t == time].max(), 100)
x2 = np.linspace(X[:, -1].min(), X[:, -1].max(), 100)
mu1 = 1 + 3 * time * (2 - time)
mu2 = 1 + 3 * T * (2 - T)
plt.plot(x1, norm.pdf(x1, loc=mu1, scale=np.sqrt(time)), 'r--', lw=1, label=f'theoretical $t={time}$')
plt.plot(x2, norm.pdf(x2, loc=mu2, scale=np.sqrt(T)), 'g--', lw=1, label=f'theoretical $t={T}$')
plt.xlabel('$X$')
plt.legend()

# sample mean
plt.subplot(2, 2, 3)
plt.plot(t, X.mean(axis=0), label='sample mean')
plt.plot(t, 1 + 3 * t * (2 - t), 'r--', label='theoretical mean')
plt.xlabel('$t$')
plt.ylabel('sample mean')
plt.legend()

# sampel standard deviation
plt.subplot(2, 2, 4)
plt.plot(t, X.std(axis=0), label='sample std')
plt.plot(t, np.sqrt(t), 'r--', label='theoretical std')
plt.xlabel('$t$')
plt.ylabel('sample standard deviation')
plt.legend()

plt.tight_layout()
plt.show()
```
